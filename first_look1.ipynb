```python
import pandas as pd
df23 = pd.read_csv("data/2023.csv")
print(df23.shape) 
```

```python
df23.info()
```

```python
df23.isna().sum()
```

```python

fa_to_en = str.maketrans("۰۱۲۳۴۵۶۷۸۹", "0123456789")


df23["price_toman"] =
(
    df23["Price"]
    .str.translate(fa_to_en)                   
    .str.replace("تومان", "", regex=False)         
    .str.replace("٬", "", regex=False)            
    .str.replace(",", "", regex=False)            
    .str.strip()                                   
)
df23["price_toman"] = 
pd.to_numeric(df23["price_toman"], errors="coerce")

df23["area_m2"] = 
pd.to_numeric(df23["Area"].astype(str).str.translate(fa_to_en), errors="coerce")


print("قیمت‌های غیرقابل تبدیل (مثلاً توافقی):", df23["price_toman"].isna().sum())
df23[["Price", "price_toman", "area_m2"]].head()
```

```python
print("تعداد آگهی با قیمت معتبر:", df23["price_toman"].notna().sum())
print("میانگین قیمت: {:,.0f} تومان".format(df23["price_toman"].mean()))
print("میانه قیمت:   {:,.0f} تومان".format(df23["price_toman"].median()))
```
